#Labour Market Analytics: A PLFS Data Analysis Using Python

#Objective


1. To analyze year-wise trends (2018–2024) in the number of persons, households, and blocks surveyed to identify changes in survey coverage over time.

2. To compare state-wise survey coverage by identifying states with the highest and lowest numbers of surveyed households and persons.

3. To examine quarter-wise (seasonal) survey coverage across the four PLFS survey periods (April–June, July–September, October–December, and January–March).

4. To analyze the demographic distribution of surveyed persons by age groups  and gender (Male/Female).

5. To identify state-wise and year-wise patterns, trends, and outliers in survey coverage using comparative analysis and visualizations.

6. To derive meaningful insights and present the findings through clear visualizations, supporting a better understanding of the PLFS survey coverage and data distribution.

Domain: Labour Market / Employment Statistics / Government Survey

```python
# Standardize text columns
df['state_name'] = df['state_name'].str.strip().str.title()
df['quarter'] = df['quarter'].str.strip()
```

```python
# Fix data types ---
count_cols = [
    'num_blocks_surveyed', 'households',
    'males_age_0_4_years', 'females_age_0_4_years', 'persons_age_0_4_years',
    'males_age_5_14_years', 'females_age_5_14_years', 'persons_age_5_14_years',
    'males_age_15_29_years', 'females_age_15_29_years', 'persons_age_15_29_years',
    'males_age_15_59_years', 'females_age_15_59_years', 'persons_age_15_59_years',
    'males_age_15_above_years', 'females_age_15_above_years', 'persons_age_15_above_years',
    'males_all_ages', 'females_all_ages', 'persons_all_ages'
]

df[count_cols] = df[count_cols].fillna(0)
df[count_cols] = df[count_cols].astype(int)
```

```python
# Display the data type of each column
print(df.dtypes)
```

**Feature Engineering**

```python
 # Flag aggregate rows (All-India) ---
df['is_all_india'] = (
    df['state_name'].str.strip().str.lower() == 'all-india'
)

df.head()
```

```python
#  Create Period Features ---
quarter_start_month = {
    'January-March': 1,
    'April-June': 4,
    'July-September': 7,
    'October-December': 10
}

df['quarter_start_month'] = df['quarter'].map(quarter_start_month)

df['period_date'] = pd.to_datetime({
    'year': df['year'],
    'month': df['quarter_start_month'],
    'day': 1
})

fiscal_quarter = {
    'January-March': 'Q4',
    'April-June': 'Q1',
    'July-September': 'Q2',
    'October-December': 'Q3'
}

df['fiscal_quarter'] = df['quarter'].map(fiscal_quarter)

df['period_label'] = (
    df['year'].astype(str) + "-" + df['fiscal_quarter']
)

df.head()
```

```python
#  Household & Survey Features ---
df['persons_per_household'] = (
    df['persons_all_ages'] / df['households']
).round(2)

df['households_per_block'] = (
    df['households'] / df['num_blocks_surveyed']
).round(2)

df['persons_per_block'] = (
    df['persons_all_ages'] / df['num_blocks_surveyed']
).round(2)

df.head()
```

```python
#  Male/Female Percentage ---
df['male_share_pct'] = (
    df['males_all_ages'] /
    df['persons_all_ages'] * 100
).round(2)

df['female_share_pct'] = (
    df['females_all_ages'] /
    df['persons_all_ages'] * 100
).round(2)
df.head()
```

```python
df['share_working_age_15_59'] = np.where(
    df['persons_all_ages'] > 0,
    (df['persons_age_15_59_years'] / df['persons_all_ages']) * 100,
    np.nan
).round(2)
df.head()
```

```python

# Create Sex Ratio (Females per 1000 Males)
df['sex_ratio_all_ages'] = np.where(
    df['males_all_ages'] > 0,
    (df['females_all_ages'] / df['males_all_ages']) * 1000,
    np.nan
).round(2)

# View the new column
print(df[['males_all_ages',
          'females_all_ages',
          'sex_ratio_all_ages']].head())
```

# **Statistical analysis & Visualizations**


```python
# MEASURES OF CENTRAL TENDENCY
df_states = df[df['is_all_india'] == False].copy()   # exclude All-India aggregate

# Column(s) of interest for this analysis
col = 'persons_all_ages'

mean_val = df_states[col].mean()
median_val = df_states[col].median()
mode_val = df_states[col].mode()[0]

print("Measures of Central Tendency:", col,"::")
print(f"Mean   : {mean_val:.2f}")
print(f"Median : {median_val:.2f}")
print(f"Mode   : {mode_val:.2f}")

```

```python
#Chart: Distribution with mean/median/mode marked
plt.figure(figsize=(9, 5))
sns.histplot(df_states[col], kde=True, color='skyblue', bins=30)
plt.axvline(mean_val, color='red', linestyle='--', label=f'Mean = {mean_val:.0f}')
plt.axvline(median_val, color='green', linestyle='--', label=f'Median = {median_val:.0f}')
plt.axvline(mode_val, color='purple', linestyle='--', label=f'Mode = {mode_val:.0f}')
plt.title(f'Distribution of {col} with Central Tendency Measures')
plt.xlabel(col)
plt.ylabel('Frequency')
plt.legend()
plt.tight_layout()
plt.savefig('central_tendency_chart.png', dpi=120)
plt.show()
```

**Interpretation**

1. The mean (5939) is slightly higher than the median (5786), indicating that the distribution is positively (right) skewed. This suggests that a few states or observations with very high numbers of surveyed persons are increasing the average.

2. The mode (3281) is much lower than both the mean and median, showing that the most frequently occurring value is lower than the overall average. This indicates that many states have relatively smaller survey counts, while fewer states have much larger counts.

3. The histogram shows that the data is not perfectly symmetric and exhibits variability across observations. The spread of values and the long right tail indicate differences in survey coverage among states, with some states having substantially higher numbers of persons surveyed than others.

```python
# VARIANCE AND STANDARD DEVIATION
variance_val = df_states[col].var()
std_val = df_states[col].std()
cv_val = (std_val / mean_val) * 100

print("\n=== Measures of Dispersion:", col, "===")
print(f"Variance             : {variance_val:.2f}")
print(f"Standard Deviation   : {std_val:.2f}")
print(f"Coefficient of Var % : {cv_val:.2f}%")
```

```python
#SKEWNESS AND KURTOSIS
skewness_val = df_states[col].skew()
kurtosis_val = df_states[col].kurt()

print("\n=== Skewness & Kurtosis:", col, "===")
print(f"Skewness : {skewness_val:.2f}  "
      f"({'Right-skewed' if skewness_val > 0 else 'Left-skewed' if skewness_val < 0 else 'Symmetric'})")
print(f"Kurtosis : {kurtosis_val:.2f}  "
      f"({'Leptokurtic (heavy tails)' if kurtosis_val > 0 else 'Platykurtic (light tails)' if kurtosis_val < 0 else 'Mesokurtic (normal-like)'})")
```

```python
# Chart: Skewness visualization ---
plt.figure(figsize=(9, 5))
sns.histplot(df_states[col], kde=True, color='orange', bins=30)
plt.title(f'Skewness = {skewness_val:.2f} | Kurtosis = {kurtosis_val:.2f}')
plt.xlabel(col)
plt.tight_layout()
plt.savefig('skew_kurtosis_chart.png', dpi=120)
plt.show()
```

**Interpretation**

1. The skewness value of 0.33 indicates that the distribution is slightly positively (right) skewed. This means that most observations are concentrated around the lower and middle values, with a few higher values extending the right tail of the distribution.

2. The kurtosis value of -0.50 indicates a platykurtic distribution. Compared to a normal distribution, the data has a flatter peak and lighter tails, suggesting fewer extreme values or outliers.

3. Overall, the distribution is fairly close to normal with only a slight right skew and moderate spread. This suggests that the persons_all_ages data is reasonably balanced and suitable for further statistical analysis and visualization.

#**Multivariate Analysis**

Multivariate Analysis examines the relationship among three or more variables simultaneously to identify patterns, trends, and interactions using techniques such as pair plots, grouped analysis, pivot tables, heatmaps, and advanced visualizations.

**Pairplot (Households, Persons, Blocks, Persons per Household, Persons per Block)**

```python

pair_cols = [
    'households',
    'persons_all_ages',
    'num_blocks_surveyed',
    'persons_per_household',
    'persons_per_block'
]

g = sns.pairplot(
    df[pair_cols],
    diag_kind='hist',
    corner=True,
    height=3.5
)

# Increase axis label size
for ax in g.axes.flatten():
    if ax is not None:
        ax.set_xlabel(ax.get_xlabel(), fontsize=12)
        ax.set_ylabel(ax.get_ylabel(), fontsize=12)
        ax.tick_params(axis='both', labelsize=10)
        ax.xaxis.label.set_rotation(15)

plt.show()
```

The pair plot provides a comprehensive view of the distribution of each survey variable (diagonal histograms) and the relationships between pairs of variables (scatter plots). It helps identify patterns, trends, correlations, and potential outliers in the PLFS dataset.

**Interpretation**
1. The diagonal histograms show that households, persons surveyed, and blocks surveyed are right-skewed, indicating that most survey observations have moderate values, while a few observations (such as All-India aggregates) have much larger values.
2. The scatter plots between households, persons surveyed, and blocks surveyed display a strong positive linear relationship. As the number of surveyed blocks increases, the number of households and persons surveyed also increases.
3. Persons per household is concentrated between 3.0 and 4.8, showing relatively little variation across survey observations.
4. There is a strong positive relationship between persons per household and persons per block, indicating that blocks with larger average household sizes also tend to have more surveyed persons.
5. A few observations are clearly separated from the main cluster, representing large aggregate records (such as All-India totals) or other high-value observations.
